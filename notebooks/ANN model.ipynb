```python
import pandas as pd
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import BatchNormalization, Dense
```

```python
train_path = "/content/training_dataset.csv"
val_path = "/content/validation_dataset.csv"
```

```python
training_dataset = pd.read_csv(train_path)
validation_dataset = pd.read_csv(val_path)
```

```python
x_train = training_dataset.iloc[:, :11]
y_train =  training_dataset.iloc[:, 11]
```

```python
x_val = validation_dataset.iloc[:, :11]
y_val = validation_dataset.iloc[:, 11]
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_auc_score
from sklearn.metrics import confusion_matrix
```

```python
def metrics(y_test, pred):
    metrics = {}
    metrics["accuracy_score"] = accuracy_score(y_test, pred)
    metrics["precision_score"] = precision_score(y_test, pred)
    metrics["recall_score"] = recall_score(y_test, pred)
    metrics["f1_score"] = f1_score(y_test, pred)
    metrics["roc_auc_score"] = roc_auc_score(y_test, pred)
    metrics["confusion_matrix"] = confusion_matrix(y_test, pred)

    return metrics
```

```python
report = {}
```

```python
x_train.shape
```

```python
model = Sequential()
model.add(Dense(1024, input_dim = (11), activation = "relu", kernel_regularizer="l2"))
model.add(BatchNormalization())
model.add(Dense(512, activation = "relu", kernel_regularizer="l2"))
model.add(BatchNormalization())
model.add(Dense(1, activation = "sigmoid", kernel_regularizer="l2"))
```

```python
model.summary()
```

```python
model.compile(optimizer="adam", loss = "binary_crossentropy", metrics = ["accuracy"])
```

```python
model.fit(x_train, y_train, epochs = 100)
```

```python
pred = model.predict(x_val)
```

```python
final_pred = (pred > 0.26)
report["ann"] = metrics(y_val, final_pred)
report["ann"]
```

```python
testing_dataset = pd.read_csv("/content/testing_dataset.csv")
x_test = testing_dataset.iloc[:, :11]
y_test =  testing_dataset.iloc[:, 11]
```

```python
predict = model.predict(x_test)
final_pred = (predict > 0.26)
metrics(y_test, final_pred)
```
